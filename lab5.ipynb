```python
import pandas as pd
import numpy as np

df = pd.read_csv("/Users/immd/Downloads/Lab5ML/ house_prices.csv")

df.head()
```

```python
# This cell helps us understand the dataset structure by checking

df.shape
df.columns
df.info()
df.describe()
```

```python
# This cell checks for missing (null) values in each column.

df.isnull().sum().sort_values(ascending=False)
```

```python
# This cell removes columns that are not useful for machine learning

df = df.drop(columns=["Index","Title","Description"], errors='ignore')
```

```python
# This cell converts the price column into a numeric format

df["Price (in rupees)"] = pd.to_numeric(df["Price (in rupees)"], errors="coerce")
```

```python
# This cell removes rows that contain missing values

df = df.dropna()
```

```python
# This cell converts categorical columns into numerical values

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in df.select_dtypes(include='object').columns:
    df[col] = le.fit_transform(df[col].astype(str))
```

```python
# This cell removes rows with missing values to avoid errors during model training

df = df.dropna()
```

```python
# This cell ensures the target column (Price) is numeric

df["Price (in rupees)"] = pd.to_numeric(df["Price (in rupees)"], errors="coerce")
```
